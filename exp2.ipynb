```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("Iris.csv")

print("First 5 rows:")
display(df.head())

print("\nDataset Information:")
print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nColumn Names:")
print(df.columns.tolist())
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["SepalLengthCm"], bins=10, edgecolor="black")

plt.xlabel("Sepal Length (cm)")
plt.ylabel("Frequency")
plt.title("Histogram of Sepal Length")
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

for species in df["Species"].unique():
    data = df[df["Species"] == species]
    plt.scatter(
        data["SepalLengthCm"],
        data["SepalWidthCm"],
        label=species
    )

plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.title("Scatter Plot of Sepal Length vs Sepal Width")
plt.legend()
plt.grid(True)

plt.show()
```

```python
species_count = df["Species"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    species_count,
    labels=species_count.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Distribution of Iris Species")
plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    df.index,
    df["SepalLengthCm"],
    marker="o"
)

plt.xlabel("Data Point")
plt.ylabel("Sepal Length (cm)")
plt.title("Line Graph of Sepal Length")
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(10, 6))

df[
    ["SepalLengthCm",
     "SepalWidthCm",
     "PetalLengthCm",
     "PetalWidthCm"]
].boxplot()

plt.xlabel("Features")
plt.ylabel("Measurement (cm)")
plt.title("Box Plot of Iris Features")
plt.grid(True)

plt.show()
```

```python
average_values = df[
    ["SepalLengthCm",
     "SepalWidthCm",
     "PetalLengthCm",
     "PetalWidthCm"]
].mean()

plt.figure(figsize=(8, 5))

plt.bar(
    average_values.index,
    average_values.values
)

plt.xlabel("Features")
plt.ylabel("Average Value (cm)")
plt.title("Average Value of Iris Features")

plt.xticks(rotation=20)
plt.grid(axis="y")

plt.show()
```
