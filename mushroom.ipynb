```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("mushrooms.csv")

```

```python
print(df.head())

print("\nNumber of Rows:", len(df))
print("Number of Columns:", len(df.columns))
```

```python
target = df.columns[0]
features = df.columns[1:]

print("\nTarget Column:")
print(target)

print("\nInput Features:")
print(list(features))
```

```python
print("\nEnter values for prediction:")

user = {}

for feature in features:
    value = input(feature + ": ")
    user[feature] = value
```

```python
print("\n===== CLASS COUNT =====")

class_count = df[target].value_counts()

print(class_count)

total_records = len(df)
```

```python
print("\n===== PROBABILITY CALCULATION =====")

final_probability = {}

for cls in class_count.index:

    print("\n--------------------------------")
    print("Class:", cls)

    # Prior Probability
    prior = class_count[cls] / total_records

    print("Prior Probability =", prior)

    probability = prior

    class_data = df[df[target] == cls]

    for feature in features:

        match = len(class_data[class_data[feature] == user[feature]])

        total_class = len(class_data)

        unique_values = df[feature].nunique()

        # Laplace Smoothing
        conditional = (match + 1) / (total_class + unique_values)

        print("P(", feature, "=", user[feature], "|", cls, ") =", conditional)

        probability = probability * conditional

    final_probability[cls] = probability

    print("Final Probability =", probability)
```

```python
for cls in final_probability:
    print(cls, ":", final_probability[cls])
```

```python

prediction = max(final_probability, key=final_probability.get)

print("\n===== PREDICTION =====")
print("Predicted Class =", prediction)

if prediction == 'e':
    print("Result: Edible Mushroom")
elif prediction == 'p':
    print("Result: Poisonous Mushroom")
```
